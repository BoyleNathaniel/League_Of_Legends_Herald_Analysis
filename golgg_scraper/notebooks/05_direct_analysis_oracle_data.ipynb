```python
# Convert date column to datetime
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Filter for the main Worlds competition
worlds_df = df[df["league"] == "WLDs"].copy()

print("Total rows in original dataset:", len(df))
print("Total Worlds rows:", len(worlds_df))
print("Unique Worlds games:", worlds_df["gameid"].nunique())

print("\nWorlds date range:")
print(worlds_df["date"].min())
print(worlds_df["date"].max())

print("\nWorlds split values:")
print(sorted(worlds_df["split"].dropna().unique()))

print("\nWorlds teams:")
print(sorted(worlds_df["teamname"].dropna().unique()))

print("\nFirst 20 Worlds game IDs:")
print(worlds_df["gameid"].dropna().unique()[:20])

# Check how many rows are recorded for each game
game_row_counts = (
    worlds_df.groupby("gameid").size().value_counts().sort_index()
)

print("\nNumber of games by row count:")
display(game_row_counts.to_frame("number_of_games"))
```

```python
# Check whether split is entirely empty
print("Missing split values:", worlds_df["split"].isna().sum())
print("Total Worlds rows:", len(worlds_df))

# Inspect playoffs values
print("\nPlayoffs values:")
print(worlds_df["playoffs"].value_counts(dropna=False))

# Inspect game values
print("\nGame values:")
print(worlds_df["game"].value_counts(dropna=False).head(30))

# Inspect participant positions
print("\nParticipant positions:")
print(worlds_df["position"].value_counts(dropna=False))
```

```python
# Count rows per game
game_row_counts = (
    worlds_df.groupby("gameid")
    .size()
    .value_counts()
    .sort_index()
)

display(game_row_counts.to_frame("number_of_games"))
```

```python
# Keep only team-level records
worlds_team_rows = worlds_df[
    worlds_df["position"] == "team"
].copy()

# Select useful columns for match inspection
match_columns = [
    "gameid",
    "date",
    "game",
    "teamname",
    "side",
    "result",
    "league",
    "year",
    "patch",
    "gamelength"
]

worlds_team_rows = worlds_team_rows[match_columns]

# Sort chronologically
worlds_team_rows = worlds_team_rows.sort_values(
    ["date", "gameid", "side"]
)

# Display the first 30 records
display(worlds_team_rows.head(30))
```

```python
# Confirm there are exactly two team records per game
print(
    worlds_team_rows.groupby("gameid").size().value_counts()
)

# Check whether each game has two distinct teams
print(
    "Games with exactly two distinct teams:",
    (worlds_team_rows.groupby("gameid")["teamname"].nunique() == 2).sum()
)

print(
    "Total unique games:",
    worlds_team_rows["gameid"].nunique()
)
```

```python
# Display the number of games recorded per date
worlds_df.groupby(
    worlds_df["date"].dt.date
)["gameid"].nunique().sort_index()
```

```python
# Filter Worlds matches from October 15, 2026 onward
worlds_df = worlds_df[
    worlds_df["date"] >= "2026-10-15"
].copy()

print("Rows:", len(worlds_df))
print("Unique games:", worlds_df["gameid"].nunique())
print("First date:", worlds_df["date"].min())
print("Last date:", worlds_df["date"].max())
```

```python
# Get team-level records for October 23
oct23_df = worlds_df[
    (worlds_df["date"].dt.date == pd.Timestamp("2026-10-23").date()) &
    (worlds_df["position"] == "team")
].copy()

# Show the matches on that date
display(
    oct23_df[
        ["gameid", "date", "game", "teamname", "side",
         "result", "patch", "gamelength"]
    ].sort_values(["date", "gameid", "side"])
)
```

```python
# Find games containing both Gen.G and Invictus Gaming
opening_match = (
    oct23_df.groupby("gameid")
    .filter(
        lambda group:
            {"Gen.G", "Invictus Gaming"}.issubset(
                set(group["teamname"])
            )
    )
)

display(opening_match)
```

```python
# Ensure date is in datetime format
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Filter for Worlds matches from the first Swiss Stage date onward
worlds_swiss_onward = df[
    (df["league"] == "WLDs") &
    (df["date"] >= "2026-10-23")
].copy()

# Summary
print("Total rows:", len(worlds_swiss_onward))
print("Unique games:", worlds_swiss_onward["gameid"].nunique())
print("First date:", worlds_swiss_onward["date"].min())
print("Last date:", worlds_swiss_onward["date"].max())

print("\nPatches:")
print(sorted(worlds_swiss_onward["patch"].dropna().unique()))
```

```python
# Count unique games per date
games_per_date = (
    worlds_swiss_onward
    .groupby(worlds_swiss_onward["date"].dt.date)["gameid"]
    .nunique()
)

display(games_per_date)
```

```python
from pathlib import Path

# Find the project root directory
current_dir = Path.cwd()
project_root = None

for folder in [current_dir, *current_dir.parents]:
    if (folder / "data" / "processed").exists():
        project_root = folder
        break

if project_root is None:
    raise FileNotFoundError("Could not locate the data/processed folder")

# Define the processed data directory
processed_dir = project_root / "data" / "processed"

print("Project root:", project_root)
print("Processed directory:", processed_dir)
```

```python
# Save the filtered Worlds dataset
output_path = processed_dir / "worlds_swiss_onward.csv"

worlds_swiss_onward.to_csv(output_path, index=False)

print(f"Saved dataset to: {output_path}")
```
